```python
# generate a parameter csv file which will be read by workflow codes

# Model parameters including:
#     - FC: Max soil moisture storage, field capacity.
#     - beta: Shape coefficient governing fate of water input to soil moisture storage.
#     - LP: Threshold for reduction of evapotranspiration.
#     - SFCF: Snowfall correction factor.
#     - TT: Threshold temperature.
#     - CFMAX: Degree-day factor.
#     - k0: Recession constant (upper storage, near surface).
#     - k1: Recession constant (upper storage).
#     - k2: Recession constant (lower storage).
#     - UZL: Threshold for shallow storage.
#     - PERC: Percolation, max flow from upper to lower storage.
```

```python
import pandas as pd

# Define parameter bounds
parameter_bounds = {'FC': [50, 1000], 
                    'beta': [1.0, 6.0],
                    'LP': [0.2, 1],
                    'SFCF': [0.5, 5], 
                    'TT': [-2.5, 2.5],
                    'CFMAX': [0.5, 10],
                    'k0': [0.05, 0.9],
                    'k1': [0.01, 0.5],
                    'k2': [0.001, 0.2],
                    'UZL': [0, 100],
                    'PERC': [0, 100],
                   }

routing_parameter_bounds = {'MAXBAS': [1, 15]}

# Combine all parameter bounds
all_bounds = {**parameter_bounds, **routing_parameter_bounds}

# Define methods for each parameter
methods = {'parTT': 'Additive'}  # parTT is Additive, others are Multiplicative by default
default_method = 'Multiplicative'

# Generate the DataFrame
data = []
for param, bounds in all_bounds.items():
    lower, upper = bounds
    default = (lower + upper) / 2  # Calculate the default as the mean of bounds
    method = methods.get(param, default_method)  # Use specific method if defined, else default
    data.append({
        'Parameter': param,
        'Default': default,
        'Lower': lower,
        'Upper': upper,
        'Method': method
    })

# Create the DataFrame
df = pd.DataFrame(data)

# Display the DataFrame
print(df)

# Save the DataFrame to a CSV file (optional)
df.to_csv('HBVKM_param_bounds.csv', index=False)
```
