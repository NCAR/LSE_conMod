```python
import pandas as pd

# # Updated parameter bounds
# parameter_bounds = {
#     'parBETA': [0.5, 10.0],  # Expanded range for steeper soil water release
#     'parFC': [20, 1000],     # Lower minimum for reduced soil retention capacity
#     'parK0': [0.05, 1.5],    # Increased upper bound for faster near-surface runoff
#     'parK1': [0.01, 0.5],    # Kept unchanged
#     'parK2': [0.001, 0.2],   # Kept unchanged
#     'parLP': [0.2, 1],       # Kept unchanged
#     'parPERC': [0, 20],      # Increased upper bound for greater percolation
#     'parUZL': [0, 50],       # Reduced upper bound for quicker activation of surface flow
#     'parTT': [-2.5, 2.5],    # Kept unchanged
#     'parCFMAX': [0.5, 15],   # Increased upper bound for faster snowmelt
#     'parCFR': [0, 0.1],      # Kept unchanged
#     'parCWH': [0, 0.2],      # Kept unchanged
#     'parBETAET': [0.3, 5],   # Kept unchanged
#     'parC': [0, 1]           # Kept unchanged
# }

# # Updated routing parameter bounds
# routing_parameter_bounds = {
#     'rout_a': [0, 5.0],  # Increased upper bound for sharper peaks in routed flow
#     'rout_b': [0, 10.0]  # Increased upper bound for sharper peaks in routed flow
# }


# Expanded parameter bounds
parameter_bounds = {
    'parBETA': [0.1, 20.0],   # Significantly expanded for steep/nonlinear soil water release
    'parFC': [10, 2000],      # Extended to accommodate very small or large soil capacities
    'parK0': [0.01, 2.0],     # Higher potential near-surface runoff rates
    'parK1': [0.001, 1.0],    # Broader range for upper storage recession
    'parK2': [0.0001, 0.5],   # Broader range for lower storage recession
    'parLP': [0.1, 1.5],      # Allowing for more flexible evapotranspiration thresholds
    'parPERC': [0, 50],       # Greatly increased for large percolation fluxes
    'parUZL': [0, 200],       # Expanded to allow a higher threshold for shallow storage activation
    'parTT': [-5.0, 5.0],     # Expanded for more diverse temperature thresholds
    'parCFMAX': [0.1, 30],    # Significantly increased for extreme snowmelt conditions
    'parCFR': [0, 0.5],       # Increased refreezing coefficients
    'parCWH': [0, 0.5],       # Increased water holding capacity
    'parBETAET': [0.1, 10],   # Expanded for diverse evapotranspiration behavior
    'parC': [0, 2]            # Extended to consider higher groundwater fluxes
}

# Expanded routing parameter bounds
routing_parameter_bounds = {
    'rout_a': [0, 10.0],  # Allowing more pronounced routing effects
    'rout_b': [0, 20.0]   # Increased to account for sharper peaks and larger basin response
}



# Combine all parameter bounds
all_bounds = {**parameter_bounds, **routing_parameter_bounds}

# Define methods for each parameter
methods = {'parTT': 'Additive'}  # parTT is Additive, others are Multiplicative by default
default_method = 'Multiplicative'

# Generate the DataFrame
data = []
for param, bounds in all_bounds.items():
    lower, upper = bounds
    default = (lower + upper) / 2  # Calculate the default as the mean of bounds
    method = methods.get(param, default_method)  # Use specific method if defined, else default
    data.append({
        'Parameter': param,
        'Default': default,
        'Lower': lower,
        'Upper': upper,
        'Method': method
    })

# Create the DataFrame
df = pd.DataFrame(data)

# Display the DataFrame
print(df)

# Save the DataFrame to a CSV file (optional)
df.to_csv('HBV_v1.1_param_bounds_EXP.csv', index=False)
```
