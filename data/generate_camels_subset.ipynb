```python
import os, sys, json
import numpy as np
```

```python
file = '/glade/campaign/cgd/tss/people/guoqiang/HBV_CAMELS/data_Yalan/Sub531ID.txt'

with open(file, 'r') as fp:
    camels531 = json.load(fp)
```

```python
# generate a few subsets: 100, 200, 300, 400, 500

camels531 = np.array(camels531)

for n in [100, 200, 300, 400, 500]:
    # Select n evenly spaced elements
    index = np.linspace(0, len(camels531) - 1, n).astype(int)
    camelsn = camels531[index].tolist()  # Convert to Python list with native types

    # Save to JSON file
    outfile = f'/glade/campaign/cgd/tss/people/guoqiang/HBV_CAMELS/data_Yalan/subsetID/Sub{n}ID.txt'
    with open(outfile, 'w') as f:
        json.dump(camelsn, f)
```
